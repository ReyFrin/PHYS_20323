### Exercise 2

Panel 1) GLON vs. GLAT

Panel 2) pmRA* vs pmDE

Panel 3) Rgc vs. Z

panel 4) AgeNN vs. DistPc

Give the sizes of the points by 'r50' and color the points by AgeNN. 

**Make sure to label ALL your axis and include a colorbar.**<br>


```python
import matplotlib.pyplot as plt
import numpy as np

data = np.genfromtxt("CG20.csv", delimiter=",", names=True)

# I looked this part up
data.dtype.names = [name.strip() for name in data.dtype.names]

colors = data["AgeNN"]
sizes = data["r50"] * 300  

fig, axs = plt.subplots(2, 2, figsize=(14, 11))

# I looked this up too
axes = axs.flatten()

scatter1 = axes[0].scatter(data["GLON"],data["GLAT"],s=sizes,c=colors,cmap="viridis",alpha=0.75,edgecolors="black",)
axes[0].set_xlabel("GLON")
axes[0].set_ylabel("GLAT")
axes[0].set_title("Panel 1) GLON vs. GLAT")
fig.colorbar(scatter1, ax=axes[0], label="AgeNN")

scatter2 = axes[1].scatter(data["pmRA"],data["pmDE"],s=sizes,c=colors,cmap="viridis",alpha=0.75,edgecolors="black",)
axes[1].set_xlabel("pmRA*")
axes[1].set_ylabel("pmDE")
axes[1].set_title("Panel 2) pmRA* vs. pmDE")
fig.colorbar(scatter2, ax=axes[1], label="AgeNN")

scatter3 = axes[2].scatter(data["Rgc"],data["Z"],s=sizes,c=colors,cmap="viridis",alpha=0.75,edgecolors="black",)
axes[2].set_xlabel("Rgc")
axes[2].set_ylabel("Z")
axes[2].set_title("Panel 3) Rgc vs. Z")
fig.colorbar(scatter3, ax=axes[2], label="AgeNN")

scatter4 = axes[3].scatter(data["AgeNN"],data["DistPc"],s=sizes,c=colors,cmap="viridis",alpha=0.75,edgecolors="black",)
axes[3].set_xlabel("AgeNN")
axes[3].set_ylabel("DistPc")
axes[3].set_title("Panel 4) AgeNN vs. DistPc")
fig.colorbar(scatter4, ax=axes[3], label="AgeNN")

plt.show()
```
